```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/raw/movies.csv")
df
```

```python
df.info()
```

```python
df.shape
```

```python
print(df.dtypes)
```

```python
print((df.isna().sum() / df.shape[0] * 100).sort_values(ascending=False))
```

```python
sns.histplot(df["budget"], bins=50)
plt.axvline(df["budget"].mean(), color="red", label="Mean")
plt.axvline(df["budget"].median(), color="blue", label="Median")

plt.legend()
plt.show()
```

```python
df["budget"] = df["budget"].fillna(df["budget"].median())
```

```python
sns.histplot(df["gross"], bins=50)
plt.axvline(df["gross"].mean(), color="red", label="Mean")
plt.axvline(df["gross"].median(), color="blue", label="Median")

plt.legend()
plt.show()
```

```python
df["gross"] = df["gross"].fillna(df["gross"].median())
```

```python
df
```

```python
df = df.dropna()
print((df.isna().sum() / df.shape[0] * 100).sort_values(ascending=False))
```

```python
df["votes"] = df["votes"].astype("int64")
df["budget"] = df["budget"].astype("int64")
df["gross"] = df["gross"].astype("int64")
df["runtime"] = df["runtime"].astype("int64")
```

```python
df.duplicated().sum()
```

```python
df.to_csv("../data/processed_data/clean_movies.csv")
```
